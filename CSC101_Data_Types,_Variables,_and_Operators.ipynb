# **Basic output**

The print() function displays output to the user. Output is the information or result produced by a program. The **sep** and **end** parameters can be used to customize the output. sep is the short for seperator.

Multiple values, separated by commas, can be printed in the same statement. By default, each value is separated by a space character in the output. The **sep** option can be used to change this behavior.

By default, the print() function adds a newline character at the end of the output. A newline character tells the display to move to the next line. The end option can be used to continue printing on the same line.

```python
print("Hello, World!")
```

The **sep** Parameter - The sep parameter controls what gets printed between multiple items. By default, it's a single space.

```python
# Default behavior (sep=' ')
print("apple", "banana", "cherry")
# Output: apple banana cherry

# Custom separator
print("apple", "banana", "cherry", sep=", ")
# Output: apple, banana, cherry

print("2024", "12", "25", sep="-")
# Output: 2024-12-25

print("a", "b", "c", sep="")
# Output: abc

print("red", "green", "blue", sep=" | ")
# Output: red | green | blue
```

The **end** Parameter - The end parameter controls what gets printed at the end of the print() statement. By default, it's a newline character (\n).

```python
# Default behavior (end='\n')
print("Hello")
print("World")
# Output:
# Hello
# World

# Custom end character
print("Hello", end=" ")
print("World")
# Output: Hello World

print("Loading", end="...")
print("Done")
# Output: Loading...Done

# Useful for progress indicators
for i in range(5):
    print("*", end="")
print()  # Add a newline at the end
# Output: *****
```

# **Variables**

Variables are used to store data values. They act as named containers that hold a particular value or object. Unlike some other programming languages, Python does not require explicit declaration of variables or specifying their data types. You can simply assign a value to a variable, and Python will automatically determine its type based on the assigned value.
Variables allow programs to refer to values using names rather than memory locations.

***Ex: age refers to a person's age, and birth refers to a person's date of birth.***

A statement can set a variable to a value using the assignment operator (=). Note that this is different from the equal sign of mathematics.

***Ex: age = 6 or birth = "May 15". The left side of the assignment statement is a variable, and the right side is the value the variable is assigned.***

A statement can set a variable to a value using the assignment operator (=).  Ex: age = 6 or birth = "May 15". The left side of the assignment statement is a variable, and the right side is the value the variable is assigned.

```python
# Assigning values to variables
age = 25
name = "bOB"
is_student = True
pi = 3.14



# Printing the values of variables
print(age)          # Output: 25
#Note the difference when you place age in quotes
print("age")
print(name)         # Output: Alice
print(is_student)   # Output: True
print(pi)           # Output: 3.14
```

Variable naming rules
A variable name can consist of letters, digits, and underscores and be of any length. The name cannot start with a digit.

***Ex: 101class is invalid.***

Also, letter case matters. Ex: Total is different from total. Python's style guide recommends writing variable names in snake case, which is all lowercase with underscores in between each word, such as first_name or total_price.

A name should be short and descriptive, so words are preferred over single characters in programs for readability.

***Ex: A variable named count indicates the variable's purpose better than a variable named just c.***

Python has reserved words, known as keywords, which have special functions and cannot be used as names for variables (or other objects).

# **DATA TYPES**

Python supports several built-in data types that are used to represent different kinds of values. Here are some of the commonly used data types in Python.

image.png


1. **Numeric Types:**
Integer (int): Used for whole numbers, positive or negative or zero. Represents integer values like 5, -10, or 1000.

Floating point (float): Represents floating-point numbers or real numbers with decimal places like 3.14, -0.5, or 2.0.

2. **Boolean Type:**

Boolean (bool): Represents either True or False. It is often used for logical operations and control flow.

3. **Sequence Types:**
String (str): Represents a sequence of Unicode characters enclosed in single quotes (' ') or double quotes (" ").

List (list): Represents an ordered collection of elements enclosed in square brackets ([]).

```python
# int: Represents integer values
age = 25
count = -10
population = 1000000
```

```python
# float: Represents floating-point numbers.
pi = 3.14
temperature = -2.5
average = 7.89
```

```python
# bool: Represents boolean values (True or False).
is_raining = True
is_sunny = False
has_passed_exam = True
```

A string is a sequence of characters enclosed by matching single (') or double (") quotes. Ex: "Happy birthday!" and '21' are both strings.

```python
# str: Represents sequences of Unicode characters.
name = "Alice"
message = 'Hello, World!'
address = "123 Main Street"
```

```python
# list: Represents ordered collections of elements.
numbers = [1, 2, 3, 4, 5]
names = ["Alice", "Bob", "Charlie"]
mixed_list = [1, "two", 3.0, True]
```

The **type()** function in Python is used to determine the type of an object. It returns the class type of the given object.

```python
x = 42
print(type(x))  # Output: <class 'int'>

y = "Hello"
print(type(y))  # Output: <class 'str'>

z = [1, 2, 3]
print(type(z))  # Output: <class 'list'>
```

# **Basic input**
The input() function in Python is a built-in way to get data from the user via the console. It pauses your program, waits for the user to type something, and returns whatever they entered as a string

An input statement, **variable = input("prompt")**, has three parts:

1. A variable refers to a value stored in memory. In the statement above, variable can be replaced with any name the programmer chooses.

2. The input() function reads one line of input from the user. A function is a named, reusable block of code that performs a task when called. The input is stored in the computer's memory and can be accessed later using the variable.

3. A prompt is a short message that indicates the program is waiting for input. In the statement above, "prompt" can be omitted or replaced with any message.

```python
name = input("Enter your name: ")
print("Hello,", name)
```

```python
# Converting input to a number
age = int(input("How old are you? "))  # Convert string to int
print("In 10 years, you'll be", age + 10)
```

# **OPERATORS AND EXPRESSIONS**

In Python, operators are used to perform operations on values or variables. An expression is a combination of operators, values, and variables that, when evaluated, produces a result.

**Basic Arithmetic Operators**

image.png


```python
a = 10
b = 3
print(a + b)    # 13
print(a - b)    # 7
print(a * b)    # 30
print(a / b)    # 3.333...
print(a // b)   # 3
print(a % b)    # 1
print(a ** b)   # 1000
```

**Combined Assignment Operators:**

image.png



```python
x = 10
x += 5   # x = x+5
print(x)  # 15

x *= 2   # x = x * 20
print(x)  # 30

x //= 4  # x is now 7
print(x)  # 7
```

In Python, operator precedence determines the order in which operations are performed in an expression.

Operators with higher precedence are evaluated first. If operators have the same precedence, they’re evaluated left-to-right (except for exponentiation, which is right-to-left). You can always override precedence with parentheses.

**Key Rules**
1. Parentheses rule all: ( ) always takes top priority.
2. Left-to-right: Most operators at the same level go left-to-right.
3. Exception: ** (exponentiation) is right-to-left.

image.png

```python
# Example
result = 2 + 3 * 4 ** 2 - (6 / 3) % 2
print(result)  # Output: 50.0
```

# **Formatting the output**



Write a Python program that calculates the area and perimeter of a rectangle. The length and width of the rectangle should be stored in variables, and the program should output the calculated area and perimeter.

```python
# Define the length and width of the rectangle
length = 10
width = 5

# Calculate the area
area = length * width

# Calculate the perimeter
perimeter = 2 * (length + width)

# Output the results
print(f"Rectangle dimensions: {length} x {width}")
print(f"Area: {area}")
print(f"Perimeter: {perimeter}")
```

The 'f' before the opening quotation mark tells Python that this is an f-string. Inside the string, we use curly braces {} to enclose Python expressions. When the string is evaluated, Python replaces these expressions with their values.

This allows us to easily combine text and variable values in our output, making the code more readable and easier to maintain.

```python
#Format the output to 2 decimal places

num = 3.14159
print(f"The value of num is {num:.2f}")
```
